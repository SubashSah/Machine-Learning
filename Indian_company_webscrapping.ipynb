```python
import requests
import numpy as np
import pandas as pd
from bs4 import BeautifulSoup
import re
```

```python
final=pd.DataFrame()
for j in range(1,11):
    headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }
    webpage=requests.get(f'https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav&page={j}',headers=headers).text
    soup=BeautifulSoup(webpage,'lxml')
    company=soup.find_all('div',class_='companyCardWrapper')
    name=[]
    rating=[]
    reviews=[]
    type=[]
    headquarter=[]
    years_old=[]
    employees=[]
    service_type=[]
    for i in company:
        name.append(i.find('h2').text.strip())
        rating.append((i.find(class_='companyCardWrapper__companyRatingValue').text.strip()))
        reviews.append(round(float(i.find('span',class_='companyCardWrapper__ActionCount').text.strip().replace('k',''))*1000))
        entry=i.find('span',class_='companyCardWrapper__interLinking').text.strip().split('|')
        if len(entry)<5:
            entry.insert(2,np.NaN)
        if len(entry)<5:
            entry.insert(3,np.NaN)
        service_type.append((entry[0]))
        employees.append(entry[1].split('+')[0].replace('Employees','').strip())
        type.append((entry[2]))
        years_old.append((entry[3]))
        headquarter.append(entry[4].split('+')[0].strip())
    d={
        'name':name,
    'rating':rating,
    'reviews':reviews,
    'type':type,
    'headquarter':headquarter,
    'years_old':years_old,
    'employees':employees,
    'service_type':service_type
    }
    df=pd.DataFrame(d)
    final=pd.concat([final,df],ignore_index=True)

```
