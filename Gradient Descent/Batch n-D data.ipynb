```python
class BGDRegressor:
    def __init__ (self,learning_rate=0.01,epochs=100):
        self.coef_=None
        self.intercept_=None
        self.lr=learning_rate
        self.epoch=epochs
        
    def fit(self,X_train,y_train):
        self.coef_=np.ones(X_train.shape[1])
        self.intercept_=0
        for i in range(self.epoch):
            y_pred=  X_train @ self.coef_ + self.intercept_  #X_train=(n,m),coef_=(m,1)
            slope_intercept=-2*(np.mean(y_train - y_pred))
            slope_coef = -2*((y_train-y_pred)@ X_train)/X_train.shape[0]
            
            self.intercept_ = self.intercept_ - (self.lr * slope_intercept)
            self.coef_ = self.coef_ - (self.lr * slope_coef)
        print(self.intercept_,self.coef_)
        
    def predict(self,X_test):
        return X_test @ self.coef_ + self.intercept_
```

```python
bgd=BGDRegressor(learning_rate=0.08,epochs=60)
```

```python
start=time.time()
bgd.fit(X_train,y_train)
print(f'Time taken is: {time.time()-start}')
```
