```python
## Dropping application_type since it has only one unique value
df.drop(columns=["application_type"], inplace=True)
```

```python
(df.emp_length.unique())
```

```python
## Function to convert emp_length into numerical:
def convert_emp_length(len):
    if len=="10+ years":
        return 10
    elif len=="< 1 year":
        return 0.5
    else:
        return int(len.split()[0])
```

```python
df["emp_length"]=df["emp_length"].apply(convert_emp_length)
```
