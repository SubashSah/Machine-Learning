```python
import pandas as pd
import numpy as np
```

```python
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, LabelEncoder,StandardScaler,MinMaxScaler
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
```

```python
df=pd.read_csv('Indian_company.csv')
```

```python
df.sample(5)
```

```python
df1=df[['rating','reviews','type']]
```

```python
df1.sample(5)
```

```python
df1['type'].fillna(df1['type'].mode()[0],inplace=True)
```

```python
df1.sample(5)
```

```python
column_transform=ColumnTransformer([('trans_rating',StandardScaler(),['rating']),('trans_review',MinMaxScaler(),['reviews']),('trans_type',OneHotEncoder(sparse_output=False),['type'])])
```

```python
df2=column_transform.fit_transform(df1)
```

```python
transformed_feature_names = column_transform.named_transformers_['trans_type'].get_feature_names_out(['type'])
```

```python
new_columns=list(df1.columns.drop('type'))+list(transformed_feature_names)
```
