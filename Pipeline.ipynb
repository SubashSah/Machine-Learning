```python
import pandas as pd
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv('titanic_dataset/train.csv')
```

```python
df.isna().sum()
```

```python
df.sample(2)
```

```python
df.drop(columns=['PassengerId','Name', 'Ticket','Cabin'],inplace=True)
```

```python
df.sample()
```

```python
# imputation transformer
trf1=ColumnTransformer([('impute_age',SimpleImputer(strategy="median"),[3]),
                       ('impute_Embarked',SimpleImputer(strategy='most_frequent'),[7])],
                      remainder='passthrough')
```

```python
# encoding and scaling:
trf2=ColumnTransformer([('ohe_sex',OneHotEncoder(sparse=False,handle_unknown='ignore'),[2,7]),
                        ('standard_scaler',StandardScaler(),[3,6])]
                       ,remainder='passthrough')
```

```python
pipe=Pipeline([('trf1',trf1),
              ('trf2',trf2)])
```

```python
df1=pd.DataFrame(pipe.fit_transform(df))
```
